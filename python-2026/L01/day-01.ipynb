# Esempio completo

```python
import camelot
import pandas as pd

tables = camelot.read_pdf("L01\\tables.pdf", pages="1")
df = tables[0].df

df.columns = ['Name', 'KI', 'Distance', 'Speed%', 'Accel%', 'Stops%', 'Idle%'] # la prima riga diventa l'elenco dei nomi delle colonne
df = df[2:]
df = df.reset_index(drop=True)

df["Speed%"] = df["Speed%"].str.replace("%", "").astype(float)
df["Accel%"] = df["Accel%"].str.replace("%", "").astype(float)
df["Stops%"] = df["Stops%"].str.replace("%", "").astype(float)
df["Idle%"] = df["Idle%"].str.replace("%", "").astype(float)

df.info()
df.describe()

df.to_csv("L01\\dati_tabella.csv", index=False)

```
