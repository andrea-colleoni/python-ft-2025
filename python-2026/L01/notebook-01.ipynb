### Tipi di PDF

| Tipo          | Cosa contiene    | Estraibile facilmente? |
| ------------- | ---------------- | ---------------------- |
| PDF testuale  | Testo reale      | ✅ Sì                   |
| PDF scansione | Solo immagine    | ❌ Serve OCR            |
| PDF misto     | Testo + immagini | ⚠ Dipende              |

```python
from PyPDF2 import PdfReader

reader = PdfReader("document.pdf")
print(reader.pages[0].extract_text())
```

Tra `PyPDF2` e `pdfplumber`, per didattica scegli **pdfplumber**:

* più stabile
* supporta bounding box
* migliore per layout

```python
%pip install pdfplumber
```

```python
import pdfplumber

with pdfplumber.open("dummy.pdf") as pdf:
    for page in pdf.pages:
        text = page.extract_text()
        print(text)
```

```python
import pdfplumber

with pdfplumber.open("form10-k.pdf") as pdf:
    page = pdf.pages[0]
    words = page.extract_words()

    for w in words[:10]:
        print(w)
```

```python
import pdfplumber
import pandas as pd

with pdfplumber.open("tables.pdf") as pdf:
    page = pdf.pages[0]
    table = page.extract_table()

df = pd.DataFrame(table[1:], columns=table[0])
print(df)
```

# 4️⃣ OCR con pytesseract

```python
%pip install pytesseract
%pip install pdf2image
```

```python
from pdf2image import convert_from_path
import pytesseract
import os

pages = convert_from_path("scanned.pdf")
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
os.environ['TESSDATA_PREFIX'] = r"C:\Program Files\Tesseract-OCR\tessdata"

for i, page in enumerate(pages):
    text = pytesseract.image_to_string(page, lang="eng")
    print(text)
```
