```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'employee': np.random.choice(['Alice','Bob','Charlie','Diana'], size=20),
    'department': np.random.choice(['Sales','Tech','HR'], size=20),
    'month': pd.date_range('2026-01-01', periods=20, freq='M'),
    'sales': np.random.randint(50, 500, size=20),
    'hours_worked': np.random.randint(120, 200, size=20)    
}) 

df.head()
df.info(memory_usage='deep')
```

# Accesso ai dati

```python
print(df['department'])
print(df.iloc[3])
print(df.iloc[3, 2]) # riga, colonna (partendo da zero)
```

## Creazione di nuove colonne calcolate

```python
df['sales_vat'] = df['sales'] * 1.22
df.head()
```

```python
risultato = []
for x in df['sales']:
    risultato.append(x * 1.22)
    print(x)

df['sales_for'] = risultato
df.head()
```

```python
df['sales_lambda'] = df.apply(lambda row: row['sales'] * 1.22, axis=1) # axis=1 => riga per riga; axis=0 (default) => colonna per colonna
df.head()
```

```python
totali = df.apply(lambda col: col[0] * 1.22) # per colonna (es: riga dei totali)
print(totali)
```

## agg() - funzioni di aggregazione

```python
print(df.agg({
    'sales': 'mean',
    'hours_worked': 'sum',
    'employee': 'nunique'
    }))

print(df['sales'].agg('max'))
print(df['sales'].agg('min'))
print(df['sales'].agg(lambda s: s.max() - s.min()))
```

## groupby()

```python
ore = df.groupby('department').agg(media_ore = ('hours_worked', 'mean'))

print(ore.head)
print(ore.loc['HR']) # loc accede alla row per label
print(ore.iloc[0]) # iloc accede alla row per indice posizionale
df.head()
```

## merge() (join)

```python
dept = pd.DataFrame({
    'department': ['Sales', 'Tech', 'HR'],
    'manager': ['Luigi', 'Anna', 'Maria'],
    'budget': [10000, 20000, 30000]
})


df_completo = df.merge(dept, how='inner', on='department')
df_completo.head()
```
